```python
#If Condition program according to patient age

age = int(input("Enter the patient's age: "))

if age >= 0 and age <= 2:
    print("Infant")
elif age > 2 and age <= 12:
    print("Child")
elif age > 12 and age <= 19:
    print("Teenager")
elif age > 19 and age <= 60:
    print("Adult")
elif age > 60:
    print("Senior Citizen")
else:
    print("Invalid Age")



```

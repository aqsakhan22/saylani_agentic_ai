```python
# Create a shopping cart system using the following data:



# products = ["Laptop", "Mouse", "Keyboard", "Headphones", "Monitor"]

# prices = [85000, 1500, 4500, 7000, 35000]

# stock = [3, 10, 5, 0, 2]



# Tasks
# Ask the user to enter three product names and their quantities.
# Check whether each product exists and has sufficient stock.
# Calculate the total bill for valid products.
# Apply discounts:
# Above 100,000 → 20%
# Above 50,000 → 10%
# Above 20,000 → 5%
# If the customer purchases a Laptop and Mouse together, apply an additional 5% discount.
# If any selected product is out of stock, display its name and exclude it from the bill.
# If the final bill exceeds 70,000, give free delivery; otherwise, charge 500.
# Update the stock list after successful purchases.
# Display the most expensive purchased product.
# Print the final receipt with purchased items, discounts, delivery charges, and total amount.
```

```python
products = ["Laptop", "Mouse", "Keyboard", "Headphones", "Monitor"]
prices = [85000, 1500, 4500, 7000, 35000]
stock = [3, 10, 5, 0, 2]

cart = []
subtotal = 0

print("========== WELCOME TO SHOPPING CART ==========")



# Step 1: Take input for three products
for i in range(3):
    product_name = input(f"\nEnter product {i + 1}: ").strip().title()
    quantity = int(input("Enter quantity: "))

    if product_name not in products:
      print("Product does not exist!")
      continue

    print('product exist')

    # Find product index
    index = products.index(product_name)

    # Step 3: Check stock
    if stock[index] == 0:
        print(f"{product_name} is out of stock!")
        continue

    if quantity <= 0:
        print("Invalid quantity!")
        continue

    if quantity > stock[index]:
        print(f"Insufficient stock for {product_name}. Available: {stock[index]}")
        continue

    item_total= prices[index] * quantity
    subtotal+=item_total

    cart.append({
        "name": product_name,
        "quantity": quantity,
        "price": prices[index],
        "total": item_total,
        "index": index
    })

# Step 5: Apply discount based on subtotal
discount = 0
if subtotal > 100000:
    discount = subtotal * 0.20
elif subtotal > 50000:
    discount = subtotal * 0.10
elif subtotal > 20000:
    discount = subtotal * 0.05

# Step 6: Additional Laptop + Mouse discount
purchased_products = [item["name"] for item in cart]

if "Laptop" in purchased_products and "Mouse" in purchased_products:
    extra_discount = (subtotal - discount) * 0.05
else:
    extra_discount = 0


total_discount = discount + extra_discount

# Step 7: Calculate delivery charges
amount_after_discount = subtotal - total_discount

if amount_after_discount > 70000:
    delivery = 0
else:
    delivery = 500

final_total = amount_after_discount + delivery

# Step 8: Update stock after successful purchases
for item in cart:
    stock[item["index"]] -= item["quantity"]

# Step 9: Find most expensive purchased product
if cart:
    expensive_product = max(cart, key=lambda item: item["price"])
else:
    expensive_product = None

# Step 10: Print receipt
print("\n========== FINAL RECEIPT ==========")

for item in cart:
    print(
        f"{item['name']} | "
        f"Qty: {item['quantity']} | "
        f"Price: Rs. {item['price']} | "
        f"Total: Rs. {item['total']}"
    )

print("-----------------------------------")
print(f"Subtotal:          Rs. {subtotal:.2f}")
print(f"Regular Discount:  Rs. {discount:.2f}")
print(f"Extra Discount:    Rs. {extra_discount:.2f}")
print(f"Total Discount:    Rs. {total_discount:.2f}")
print(f"Delivery Charges:  Rs. {delivery:.2f}")
print(f"Final Amount:      Rs. {final_total:.2f}")

if expensive_product:
    print(
        f"Most Expensive Purchased Product: "
        f"{expensive_product['name']} "
        f"(Rs. {expensive_product['price']})"
    )
else:
    print("No products purchased.")

print("-----------------------------------")
print("Remaining Stock:", stock)
print("Thank you for shopping!")

```

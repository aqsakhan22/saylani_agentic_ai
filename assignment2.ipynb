Objective: Write a Python program that determines the electricity rate per unit based on the number of units consumed using input(), if, elif, and else.

Assignment Requirements
Ask the user to enter their name.
Ask the user to enter the number of electricity units consumed.
Use if, elif, and else to determine the rate per unit.

Electricity Rate Table
Units Consumed

Rate per Unit

0–100

Rs. 10

101–200

Rs. 15

201–300

Rs. 20

Above 300

Rs. 25



Expected Output
Enter customer name: Ahmed

Enter electricity units: 250



Customer Name: Ahmed

Units Consumed: 250

Rate Per Unit: Rs. 20

```python
name=input('enter your name')
num_of_unit=int(input('enter number of unit'))

if num_of_unit >= 0 and num_of_unit <=100:
  print('RS 10')
elif num_of_unit >= 101 and num_of_unit <=200:
    print('RS 15')
elif num_of_unit >= 201 and num_of_unit <=300:
    print('RS 20')
else :
  print('RS 25')




```
